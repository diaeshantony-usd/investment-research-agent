# 08. Brief Refiner Agent (`BriefRefiner`) Demo

### Architecture Role & Responsibilities
The **Refiner Agent** forms the optimizer side of the *evaluator-optimizer* review loop. When the Critic flags issues or unverified numbers, the Refiner:

1. Rewrites **only** the flagged sections without modifying solid content.
2. Never fabricates numbers—fetches or verifies figures against tool data.
3. Appends a structured `### Change Summary` documenting every modification.

```python
import sys
from pathlib import Path

# Resolve project src directory
project_root = Path.cwd()
for candidate in [Path.cwd(), Path.cwd().parent, Path.cwd().parent.parent]:
    if (candidate / "src").exists():
        project_root = candidate
        break
sys.path.insert(0, str(project_root / "src"))

from research_agent.agents.refiner import refiner_agent
from research_agent.interfaces import get_llm

llm = get_llm()
refiner = refiner_agent(llm=llm)
print(f"Agent initialized: {refiner.name}")
```

## Sending Flagged Issues & Executing Targeted Refinement

```python
prompt = (
    "Refine the draft brief to address these Critic issues:\n"
    "1. Clarify GAAP vs non-GAAP gross margin figures (75.1% vs 75.7%).\n"
    "2. Incorporate specific SEC 10-Q risk disclosures regarding customer concentration.\n\n"
    "Current Brief:\n"
    "# Research Brief: NVIDIA Corp. (NVDA)\n"
    "## Executive Summary\nStrong operational momentum.\n"
    "## 1. Fundamental & Earnings Performance\nRevenue $30.04B with gross margin 75%.\n"
    "## 2. Technical Profile & Market Momentum\nPrice $128.50, 14-day RSI 63.34.\n"
    "## 3. Catalysts, Macro Context & Risk Analysis\nRisks present.\n"
    "## 4. Investment Stance & Outlook\nBullish.\n"
)

print("Executing Refiner with targeted feedback...\n")
resp = refiner.run(prompt)
print("--- REFINER OUTPUT WITH CHANGE LOG ---")
print(resp.content)
```
