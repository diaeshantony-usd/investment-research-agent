# 06. Draft Writer Agent (`DraftWriter`) Demo

### Architecture Role & Responsibilities
The **Draft Writer** is responsible for institutional synthesis. It ingests findings from the specialist blackboard (`GraphState.agents`) and constructs a complete equity research brief formatted in Markdown with five mandatory sections:

1. Executive Summary
2. Fundamental & Earnings Performance
3. Technical Profile & Market Momentum
4. Catalysts, Macro Context & Risk Analysis
5. Stance & Investment Outlook

```python
import sys
from pathlib import Path

# Resolve src directory
project_root = Path.cwd().parent if Path.cwd().name == "demo" else Path.cwd()
sys.path.insert(0, str(project_root / "src"))

from research_agent.agents.draft_writer import draft_writer_agent
from research_agent.interfaces import get_llm

llm = get_llm()
draft_writer = draft_writer_agent(llm=llm)
print(f"Agent initialized: {draft_writer.name}")
```

## Ingesting Specialist Findings & Generating Institutional Draft Brief

```python
import json

specialist_findings = {
    "EarningsAnalyst": "NVDA Q2 Revenue: $30.04B (+122% YoY); Data Center $26.3B; Gross margin 75.1%; Operating income $18.64B; Free cash flow $14.5B.",
    "MarketAnalyst": "NVDA Current Price: $128.50; 50-day SMA $121.30; 200-day SMA $98.40; 14-day RSI 63.34; Beta 1.72.",
    "NewsAnalyst": "Blackwell AI processor volume shipments scheduled for Q4. Hyperscaler capex expanded over $40B. Overall sentiment bullish (0.82).",
}

prompt = (
    "User Request / Goal:\n"
    "Synthesize an institutional research brief for NVDA based on specialist research findings.\n\n"
    f"Prior Specialist Findings:\n{json.dumps(specialist_findings, indent=2)}"
)

print("Executing Draft Writer with specialist findings...\n")
resp = draft_writer.run(prompt)
print("--- GENERATED RESEARCH BRIEF ---")
print(resp.content)
```
