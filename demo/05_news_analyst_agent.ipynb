# 05. News Analyst Agent (`NewsAnalyst`) Demo

### Architecture Role & Responsibilities
The **News Analyst** scours live press releases, news feeds, and catalyst announcements. It structures headlines into categorized catalysts and classifies market sentiment using:

- `get_news_data`: News article retrieval across recent trade days with sentiment polarity scoring.

```python
import sys
from pathlib import Path

# Resolve src directory
project_root = Path.cwd().parent if Path.cwd().name == "demo" else Path.cwd()
sys.path.insert(0, str(project_root / "src"))

from research_agent.agents.news_analyst import news_analyst_agent
from research_agent.interfaces import get_llm

llm = get_llm()
news_analyst = news_analyst_agent(llm=llm)
print(f"Agent: {news_analyst.name}")
print(f"Tools: {list(news_analyst.tools.keys())}")
```

## Sending News Catalyst Retrieval Request

```python
prompt = "Synthesize recent news catalysts, media headlines, and corporate sentiment for NVDA"
print(f"Sending Input: '{prompt}'\n")

resp = news_analyst.run(prompt)
print("--- NEWS ANALYST OUTPUT ---")
print(resp.content)
```

## Inspecting Tool Execution Trace

```python
print("--- TOOL EXECUTION TRACE ---")
for step in resp.trace:
    if step.step_type == "tool_call":
        print(f"TOOL CALL   : {step.content.get('tool')}({step.content.get('arguments')})")
    elif step.step_type == "tool_result":
        preview = str(step.content)[:90].replace("\n", " ")
        print(f"OBSERVATION : {preview}...")
    elif step.step_type == "final_answer":
        print(f"FINAL ANSWER: Generated ({len(str(step.content))} chars)")
```
